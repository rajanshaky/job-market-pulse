# Job Market Pulse — Analysis Notebook
Analysis of 1,149 live job listings fetched from Adzuna API (April 2026)

## Dataset
- **Source:** Adzuna API
- **Total Jobs:** 1,149
- **Keywords:** Data Analyst, Business Analyst, Data Scientist, Power BI, SQL Analyst
- **Country:** India

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.style.use('seaborn-v0_8')
df = pd.read_csv('../../data/processed/job_listings_clean.csv')
print(f"Dataset shape: {df.shape}")
df.head() 
```

## 1. Top 10 Hiring Cities

```python
plt.figure(figsize=(10,6))
city_counts = df['city'].value_counts().head(10)
sns.barplot(x=city_counts.values, y=city_counts.index, palette='viridis')
plt.title('Top 10 Hiring Cities for Data Roles in India')
plt.xlabel('Number of Job Postings')
plt.tight_layout()
plt.show()
```

**Key Finding:** Bangalore dominates with 336 job postings — nearly double Hyderabad (157). 
"India" represents remote/pan-India roles.

## 2. Jobs by Role Type

```python
plt.figure(figsize=(8,5))
keyword_counts = df['search_keyword'].value_counts()
sns.barplot(x=keyword_counts.values, y=keyword_counts.index, palette='coolwarm')
plt.title('Job Postings by Role Type')
plt.xlabel('Number of Postings')
plt.tight_layout()
plt.show()
```

**Key Finding:** Data Analyst, Business Analyst, Data Scientist and Power BI 
roles are equally in demand at ~250 each. SQL Analyst has fewer openings (149).

## 3. Top 10 Hiring Companies

```python
plt.figure(figsize=(10,6))
company_counts = df['company'].value_counts().head(10)
sns.barplot(x=company_counts.values, y=company_counts.index, palette='magma')
plt.title('Top 10 Hiring Companies for Data Roles')
plt.xlabel('Number of Job Postings')
plt.tight_layout()
plt.show()
```

**Key Finding:** TCS leads with 43 openings — nearly double second place Turing (23). 
Mix of Indian IT giants and global firms actively hiring data professionals.

## 4. Jobs by Category

```python
plt.figure(figsize=(10,6))
cat_counts = df['category'].value_counts().head(8)
sns.barplot(x=cat_counts.values, y=cat_counts.index, palette='Blues_r')
plt.title('Job Postings by Category')
plt.xlabel('Number of Postings')
plt.tight_layout()
plt.show()
```

**Key Finding:** 87% of data role postings fall under IT Jobs category. 
Accounting & Finance (80) is a distant second — showing demand for data skills in finance.

## 5. Monthly Posting Trend

```python
plt.figure(figsize=(12,5))
monthly = df['month'].value_counts().sort_index()
plt.plot(monthly.index, monthly.values, marker='o', color='steelblue')
plt.title('Job Postings by Month')
plt.xlabel('Month')
plt.ylabel('Number of Postings')
plt.xticks(range(1,13), ['Jan','Feb','Mar','Apr','May','Jun',
                          'Jul','Aug','Sep','Oct','Nov','Dec'])
plt.tight_layout()
plt.show()
```

**Key Finding:** April spike reflects live data fetched in April 2026. 
Historical months show consistent baseline demand of 12-42 postings.
